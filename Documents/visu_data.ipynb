```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("streamlit/regularite-mensuelle-tgv-aqst.csv", sep=";")
```

```python
g= sns.pairplot(df)
for ax in g.axes.flat:
    if ax is not None:
        # Incline les étiquettes de l'axe X (horizontal)
        ax.tick_params(axis='x', labelrotation=45)
        
        # Incline les étiquettes de l'axe Y (vertical) si nécessaire
        ax.tick_params(axis='y', labelrotation=45)

# 4. Affichage du graphique
plt.tight_layout()
plt.show()
```
